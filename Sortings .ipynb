```python
a=[23,66,1,2,99,100,55]
count=0
for i in range(len(a)):
  for j in range(len(a)-1-i):
    if a[j]>a[j+1]:
      a[j],a[j+1]=a[j+1],a[j]
    count=count+1
    print(a)
    print(count)
```

```python
def sort(a):
  for i in range(len(a)):
    swapped=False
    for j in range(len(a)-1-i):
      if a[j]>a[j+1]:
         a[j],a[j+1]=a[j+1],a[j]
         swapped=True
      if not swapped:
        break
    return a
a=[23,66,1,2,99,100,55]
sort(a)
print(a)
```
